```
from google.colab import files
import pandas as pd
import duckdb

print('Selecione o CSV original de Alagoas:')
uploaded = files.upload()
origem = next(iter(uploaded))
df = pd.read_csv(origem, sep=';', encoding='latin-1', dtype=str, keep_default_na=False)
print('Linhas e colunas:', df.shape)
assert df.shape == (15353, 53), 'Verifique se o CSV é o arquivo esperado.'
saida = 'despesas_contratadas_candidatos_2026_AL.parquet'
con = duckdb.connect()
con.register('dados', df)
con.table('dados').write_parquet(saida, compression='zstd')
checagem = con.execute('SELECT COUNT(*) AS linhas FROM read_parquet(?)', [saida]).fetchone()[0]
assert checagem == len(df)
print('Parquet gerado com', checagem, 'registros.')
files.download(saida)
```
